# 00. Topic Scoping — Educational Technology

이번 실습 주제는 **교육공학**입니다.

너무 넓은 `educational technology`를 그대로 검색하지 않고, 논문 아이디어로 이어질 수 있는 하위 주제로 좁힙니다.

**Selected topic**  
`AI-supported formative feedback for self-regulated learning in educational technology`


```python
from pathlib import Path
from IPython.display import Markdown, display

DATA_DIR = Path('../data')
DATA_DIR.mkdir(parents=True, exist_ok=True)

TOPIC_KO = '교육공학에서 AI 기반 형성적 피드백이 자기조절학습에 미치는 영향'
RESEARCH_QUESTION = 'How does AI-supported formative feedback influence self-regulated learning in educational technology environments?'
QUERY = 'AI feedback self-regulated learning educational technology'
KEYWORDS = ['educational technology', 'artificial intelligence in education', 'AI feedback', 'formative feedback', 'self-regulated learning', 'learning analytics']
FROM_YEAR = 2020
TO_YEAR = 2026
SEARCH_FIELD = 'title_and_abstract'
LANGUAGE = 'en'
```

## 연구 질문 후보

실제 연구에서는 이 중 하나를 더 좁혀 선택합니다. 이번 실습은 후보 1을 기본값으로 사용합니다.


```python
candidates = [
    {
        'type': 'main practice question',
        'question': RESEARCH_QUESTION,
        'keywords': QUERY,
        'fit': 'good default for literature search and abstract drafting',
    },
    {
        'type': 'chatbot focus',
        'question': 'How do generative AI chatbots support goal setting, monitoring, and reflection in self-regulated learning?',
        'keywords': 'generative AI chatbot self-regulated learning education',
        'fit': 'narrower and easier to turn into a design-principle paper',
    },
    {
        'type': 'learning analytics focus',
        'question': 'How does learning analytics feedback affect students motivation and learning performance?',
        'keywords': 'learning analytics feedback motivation learning performance educational technology',
        'fit': 'useful if you want a data/analytics-centered paper',
    },
    {
        'type': 'design principle focus',
        'question': 'What design principles improve AI feedback systems for student self-regulation?',
        'keywords': 'AI feedback design principles self-regulated learning education',
        'fit': 'good for a conceptual or design-based research paper',
    },
]

import pandas as pd
df_candidates = pd.DataFrame(candidates)
display(df_candidates)
```

```python
out = DATA_DIR / 'topic_keywords.txt'
out.write_text(
    f'TOPIC_KO={TOPIC_KO}\n'
    f'RESEARCH_QUESTION={RESEARCH_QUESTION}\n'
    f'QUERY={QUERY}\n'
    f'KEYWORDS={"; ".join(KEYWORDS)}\n'
    f'FROM_YEAR={FROM_YEAR}\n'
    f'TO_YEAR={TO_YEAR}\n'
    f'SEARCH_FIELD={SEARCH_FIELD}\n'
    f'LANGUAGE={LANGUAGE}\n',
    encoding='utf-8',
)
print(f'Saved -> {out.resolve()}')
```
